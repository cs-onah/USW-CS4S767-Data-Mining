## Correlation between Height and weight.

```python
import numpy as np
import matplotlib.pyplot as plt

height = np.array([150, 140, 170, 184, 185, 178,161,152,170,165,188])
weight = np.array([54, 44, 67, 81, 92, 77, 65, 55, 72,68,92])

# Pearson correlation coefficient
r = np.corrcoef(height, weight)[0, 1]
print(f"Correlation (r): {r:.4f}")

# Scatter plot with best-fit line
m, b = np.polyfit(height, weight, 1)
plt.scatter(height, weight, color="blue", label="Data")
plt.plot(height, m * height + b, color="red", label=f"Best fit (r = {r:.2f})")
plt.xlabel("Height (cm)")
plt.ylabel("Weight (kg)")
plt.title("Height vs Weight")
plt.legend()
plt.grid(True)
plt.show()
```
